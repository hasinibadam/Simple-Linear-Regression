```python
# --- Task 3: Linear Regression from Scratch ---
import numpy as np
import matplotlib.pyplot as plt

# --- Step 1: Create Dummy Data ---
# Example: Predicting salary based on years of experience
X = np.array([1, 2, 3, 4, 5])   # Years of Experience
Y = np.array([30000, 35000, 40000, 45000, 50000])  # Salary in Rupees

# --- Step 2: Compute Mean ---
mean_x = np.mean(X)
mean_y = np.mean(Y)
n = len(X)

# --- Step 3: Calculate slope (m) and intercept (c) ---
numerator = np.sum((X - mean_x) * (Y - mean_y))
denominator = np.sum((X - mean_x)**2)
m = numerator / denominator
c = mean_y - (m * mean_x)

print(f"Equation of Line: y = {m:.2f}x + {c:.2f}")

# --- Step 4: Make Predictions ---
Y_pred = m * X + c

# --- Step 5: Visualization ---
plt.scatter(X, Y, color='blue', label='Actual Data')
plt.plot(X, Y_pred, color='red', label='Regression Line')
plt.xlabel("Years of Experience")
plt.ylabel("Salary (Rupees)")
plt.title("Linear Regression from Scratch")
plt.legend()
plt.show()

# --- Step 6: Evaluate Model ---
# Calculate Mean Squared Error (MSE)
mse = np.mean((Y - Y_pred)**2)
print(f"Mean Squared Error: {mse:.2f}")

# --- Step 7: Predict for New Data ---
new_exp = 6
predicted_salary = m * new_exp + c
print(f"Predicted salary for {new_exp} years of experience: ₹{predicted_salary:.2f}")
```
